```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Set seed for reproducibility
np.random.seed(42)
random.seed(42)

num_records = 1000

# Generating realistic Nigerian data options
hospitals = ['LUTH (Lagos)', 'UCH (Ibadan)', 'AKTH (Kano)', 'UNTH (Enugu)', 'NHA (Abuja)', 'Ahmadu Bello UTH', 'UBTH (Benin)', 'FMC Owerri']
zones = ['South West', 'South West', 'North West', 'South East', 'North Central', 'North West', 'South South', 'South East']
hospital_zone_map = dict(zip(hospitals, zones))

diseases = ['Malaria Severe', 'Typhoid Fever', 'Maternal Delivery', 'Hypertension Management', 'Type 2 Diabetes', 'Appendectomy', 'Road Traffic Injury']
base_costs = [15000, 12000, 45000, 25000, 30000, 120000, 85000]
disease_cost_map = dict(zip(diseases, base_costs))

statuses = ['Approved', 'Approved', 'Approved', 'Rejected', 'Under Review']
reasons = ['Incomplete Documentation', 'Exceeded Tariff Cap', 'Duplicate Claim', 'Non-Covered Treatment']

# Generate synthetic arrays
claim_ids = [f"NHIS-{i:04d}" for i in range(1001, 1001 + num_records)]
chosen_hospitals = [random.choice(hospitals) for _ in range(num_records)]
chosen_zones = [hospital_zone_map[h] for h in chosen_hospitals]
chosen_diseases = [random.choice(diseases) for _ in range(num_records)]

# Calculate varying costs and NHIS co-pay (NHIS covers 90% of primary/secondary tier treatments usually)
billed_amounts = [int(disease_cost_map[d] * random.uniform(0.85, 1.25)) for d in chosen_diseases]
nhis_coverage = [int(amt * 0.90) for amt in billed_amounts]

claim_statuses = [random.choice(statuses) for _ in range(num_records)]
rejection_reasons = [random.choice(reasons) if s == 'Rejected' else 'N/A' for s in claim_statuses]

# Generate random dates across recent years
start_date = datetime(2024, 1, 1)
date_list = [start_date + timedelta(days=random.randint(0, 900)) for _ in range(num_records)]

# Construct DataFrame
df_nhis = pd.DataFrame({
    'claim_id': claim_ids,
    'hospital_name': chosen_hospitals,
    'geopolitical_zone': chosen_zones,
    'diagnosis': chosen_diseases,
    'billed_amount_naira': billed_amounts,
    'nhis_payout_naira': nhis_coverage,
    'claim_status': claim_statuses,
    'rejection_reason': rejection_reasons,
    'submission_date': date_list
})

# Export directly to file directory
df_nhis.to_csv('nigerian_nhis_claims.csv', index=False)
print(f"Successfully generated {num_records} NHIS data rows as 'nigerian_nhis_claims.csv'!")
```
